## Task 1: While Loop & Control Statements – Number Guessing Game

**Objective:** Generate a random number from 1 to 10 and allow the user a maximum of 3 guesses.

The program uses `while`, `continue`, `break`, and `else` as required.

### Step 1: Import the `random` module

````python
import random
````

### Step 2: Generate a random number between 1 and 10

````python
# Generate random number between 1 and 10
secret_number = random.randint(1, 10)

print("A secret number has been generated between 1 and 10.")
````

### Step 3: Set the maximum number of attempts

The assignment specifies **3 attempts**.

````python
attempts = 3
````

### Step 4: Implement the guessing logic

- `while` keeps the game running while attempts remain.
- `continue` skips the rest of the loop when the guess is outside 1–10.
- `break` ends the game when the correct number is guessed.
- `else` belongs to the `while` loop and runs only when the loop finishes normally without `break`.

````python
attempts = 3

while attempts > 0:
    guess = int(input("Guess the number (between 1 and 10): "))

    # Check whether the guess is within the valid range
    if guess < 1 or guess > 10:
        print("Your guess is out of range. Please guess a number between 1 and 10.")
        continue

    # One valid guess has been used
    attempts -= 1

    # Check whether the guess is correct
    if guess == secret_number:
        print("Congratulations! You guessed the correct number.")
        break
    elif guess > secret_number:
        print("Too high. Try again.")
    else:
        print("Too low. Try again.")
else:
    print("Better luck next time!")
    print("The secret number was:", secret_number)
````

### Task 1 – Control Statements Used

| Control statement | Where it is used | Purpose |
|---|---|---|
| `while` | `while attempts > 0` | Repeats the guessing process |
| `continue` | Out-of-range guess | Skips the remaining statements for that iteration |
| `break` | Correct guess | Immediately exits the loop |
| `else` | After the `while` loop | Displays the failure message when all valid attempts are used |

**Note:** An out-of-range guess does not reduce the 3 valid guessing attempts because `continue` occurs before `attempts -= 1`.

## Task 2: For Loop – Multiplication Table Generator

**Objective:** Ask the user for a number and generate its multiplication table from 1 to 10 using a `for` loop and `range()`.

### Step 1: Prompt the user for a number

````python
number = int(input("Enter the number for which you want the multiplication table: "))
````

### Step 2: Generate the multiplication table using `for` and `range()`

````python
for i in range(1, 11):
    result = number * i
    print(f"{number} x {i} = {result}")
````


## Task 3: Function – BMI Calculator

**Formula:**

`BMI = weight (kg) / [height (m)]²`

The assignment requires a function named `calculate_bmi(weight, height)`.

### Step 1: Define the `calculate_bmi()` function

````python
def calculate_bmi(weight, height):
    bmi = weight / (height ** 2)
    return bmi
````

### Step 2: Get weight and height from the user

````python
weight = float(input("Enter your weight in kg: "))
height = float(input("Enter your height in meters: "))
````

### Step 3: Call the function and display the BMI

````python
bmi = calculate_bmi(weight, height)
print(f"Your BMI is: {bmi:.2f}")
````
